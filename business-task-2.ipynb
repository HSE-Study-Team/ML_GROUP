Для улучшения результатов модели в дальнейшем будем эксперимаентировать со списком признаков для удаления и сравнивать метрики. Сразу подготовим для этого функцию , возвращающую матрицу объектов-признаков X и таргет y.

```python
cols_to_drop = ["lifetime_value", "user_revenue", "lifetime_per_year", 'city']

def prepare_X_y(df:pd.DataFrame, drop_cols:list[str], target:str="lifetime_value"):
  y = df[target]
  X = df.drop(columns=drop_cols)

  bools = df.select_dtypes(include='bool').columns
  X[bools] = X[bools].astype(int)

  return X, y
```

## Разделение train - test

```python
X, y = prepare_X_y(df, cols_to_drop)
```

```python
X.head()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=13)
```

## Бейзлайн модели линейной регрессии

```python
from sklearn.linear_model import LinearRegression
```

```python
def linear_regressor(X_train, y_train, X_test):

  lr = LinearRegression()
  lr.fit(X_train, y_train)

  y_pred = lr.predict(X_test)
  return y_pred
```

## Бейзлайн модели KNN


```python
from sklearn.neighbors import KNeighborsRegressor
```

```python
def knn_regressor(X_train, y_train, X_test, n_neighbors=3, p=1):

  knn = KNeighborsRegressor(n_neighbors = n_neighbors, p = p)
  knn.fit(X_train, y_train)

  y_pred = knn.predict(X_test)
  return y_pred
```

## Функция подсчёта метрик


* Mean Squared Error (MSE): $\frac{1}{n}\sum\limits_i^n(\widetilde{y}_i - y_i)^2$
* Mean Absolute Error (MAE): $\frac{1}{n}\sum\limits_i^n|\widetilde{y}_i - y_i|$
* Max Error: $\max\limits_i |\widetilde{y}_i - y_i|$

где $\widetilde{y}$ — предсказанное значение, $y$ — реальное значение.

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, max_error
from sklearn.metrics import r2_score, mean_absolute_percentage_error
```

```python
def show_metrics(y_test, y_pred):

  print("MSE: ", mean_squared_error(y_test, y_pred))
  print("MAE: ", mean_absolute_error(y_test, y_pred))
  print("MAX_ERROR: ", max_error(y_test, y_pred))
  print("RMSE: ", np.sqrt(mean_squared_error(y_test, y_pred)))

  print("R2: ", r2_score(y_test, y_pred))
  print("MAPE: ", mean_absolute_percentage_error(y_test, y_pred))
```

```python
show_metrics(y_test, linear_regressor(X_train, y_train, X_test))
```

```python
show_metrics(y_test, knn_regressor(X_train, y_train, X_test))
```

Чтож, метрики у knn лучше по MAE, RMSE, R2. Значит модель описывает ценность клиента методом ближайших соседей точнее чем линейной регрессией. По крайней мере, по текущему набору признаков.

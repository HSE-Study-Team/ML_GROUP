#### Обработка пропусков

```python

def show_nan_info(data):
  nan_data = data.isna().sum()
  nan_data = nan_data[nan_data > 0].to_frame(name="nan_count")

  nan_data['nan_%'] = (nan_data["nan_count"] / data.shape[0]) * 100
  return nan_data.sort_values(by='nan_%', ascending=False)

nan_data = show_nan_info(raw_data)
nan_data
```

Все признаки float. Вероятно это связано, что к этому типу приводится при наличии Nan
- Часть признаков описвает число определенных действий. Можно предположить что Nan находится там, где этих действий 0. (то есть ничего не происходило)
- Проверим какие из этих признаков не содержат 0

```python
nan_cols_without_zero = []

for feat in nan_cols_list:
  if 0 not in raw_data[feat].values:
    nan_cols_without_zero.append(feat)
nan_cols_without_zero
```

Это кандидаты по заполнению Nan нулями
- Однако то, что они не содержат 0 это логично. Они не относятся к категории, которуя описана выше.
- Поэтому пока этот анализ результатов не дал, проверим другие стратегии по работе с пропусками в этих колонаках.


- Подумаем над тем, по каким признакам можно просто исключить строки (dropna). Логично будет это сделать посмотрев корреляцию признаков с целевой переменной, в нашем случае это Churn.

```python
import seaborn as sns

for_corr = raw_data[nan_cols_list + ['Churned']].corr()['Churned'].sort_values(ascending=False)
sns.heatmap(for_corr.to_frame(), cmap='coolwarm', annot=True)
```

```python
# да, коореляция не всегда бывает прямая, но в контесте объема датасета в 50К, считаем,
# что можно удалить пропуски в наиболее "отдаленных" от целевой переменной признаков.

cols_to_drop = ['Payment_Method_Diversity', 'Returns_Rate', 'Discount_Usage_Rate']


print("До очистки", raw_data.shape)
raw_data = raw_data.dropna(subset=cols_to_drop)
print(raw_data.shape)
```

```python
show_nan_info(raw_data.dropna(subset=cols_to_drop))
```

Вернемся к количественным признакам, в которых Nan означают отсутсвие чего-либо, тоесть их можно грубо заменить на 0.

```python
nan_to_zero_cols  = ['Social_Media_Engagement_Score',
                     'Mobile_App_Usage',
                     'Wishlist_Items',
                     'Product_Reviews_Written',
                     'Customer_Service_Calls',
                     ]
for col in nan_to_zero_cols:
  raw_data[col] = raw_data[col].fillna(0)

# смотрим текущую статистику по пропускам
show_nan_info(raw_data)
```

#### Обработка выбросов

Для обработки выбросов проанализирую то, какие значения принимают признаки, построив диаграмму

```python
import numpy as np
import matplotlib.pyplot as plt

# посмотрю на распределения всех числовых типов и опеределю, где необходимо удалять выбросы
# для наглядности распределения логарифмирую
def plot_log_boxplots(df):
    # num_df = df.select_dtypes(include=['number'])
    log_df = df.apply(lambda col: np.log(col + 1) if col.min() >= 0 else col)

    plt.figure(figsize=(15, 8))
    # log_df.boxplot()
    df.boxplot()
    plt.show()

plot_log_boxplots(raw_data[[
                     'Session_Duration_Avg',
                     'Pages_Per_Session',
                     'Age',
                     ]])
```

Видно "неадекватные" значения для age. Можно обрезать до стандартного (0,90)

```python
raw_data['Age'] = raw_data['Age'].clip(0, 90)
```

```python
# диапазон (0, 100) для признаков _Rate
plot_log_boxplots(raw_data[
    ['Email_Open_Rate']
])

# todo
raw_data['Email_Open_Rate'] = raw_data['Email_Open_Rate'].clip(0, 90)
```

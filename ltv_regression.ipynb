Наш таргет - lifetime_value - выручка от клиента. Кажется что признаки, касающиеся lifetime_per_year, user_revenue.

```python
cols_to_drop = ["lifetime_value", "user_revenue", "lifetime_per_year", 'city']

def prepare_X_y(df:pd.DataFrame, drop_cols:list[str], target:str="lifetime_value"):
  y = df[target]
  X = df.drop(columns=drop_cols)

  bools = df.select_dtypes(include='bool').columns
  X[bools] = X[bools].astype(int)

  return X, y
```

## Функция подсчёта метрик


* Mean Squared Error (MSE): $\frac{1}{n}\sum\limits_i^n(\widetilde{y}_i - y_i)^2$
* Mean Absolute Error (MAE): $\frac{1}{n}\sum\limits_i^n|\widetilde{y}_i - y_i|$
* Max Error: $\max\limits_i |\widetilde{y}_i - y_i|$

где $\widetilde{y}$ — предсказанное значение, $y$ — реальное значение.

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, max_error
from sklearn.metrics import r2_score, mean_absolute_percentage_error
```

```python
def show_metrics(y_test, y_pred):

  print("MSE: ", mean_squared_error(y_test, y_pred))
  print("MAE: ", mean_absolute_error(y_test, y_pred))
  print("MAX_ERROR: ", max_error(y_test, y_pred))
  print("RMSE: ", np.sqrt(mean_squared_error(y_test, y_pred)))

  print("R2: ", r2_score(y_test, y_pred))
  print("MAPE: ", mean_absolute_percentage_error(y_test, y_pred))
```

```python
show_metrics(y_test, linear_regressor(X_train, y_train, X_test))
```

Чтож, метрики у knn лучше по MAE, RMSE, R2. Значит модель описывает ценность клиента методом ближайших соседей точнее чем линейной регрессией. По крайней мере, по текущему набору признаков.


Также видно что MAPE имеет несвойственное значение. Пожалуй, это может быть связано с числом маленьких чисел (и даже нулей) при анализе. Проверим это, попробуем посчитать теже метрики, "отбросив" нули.

Переработаем функцию вывода метрик в виде конструктора, где требуемые метрики подаются как аргументы.

```python
metrics = ['MSE', 'MAE', 'MAX_ERROR', 'RMSE', 'R2', 'MAPE']

def show_metrics(y_test, y_pred, metrics: list[str], make_non_zero=False):
  if 'MSE' in metrics:
    print("MSE: ", mean_squared_error(y_test, y_pred))
  if 'MAE' in metrics:
    print("MAE: ", mean_absolute_error(y_test, y_pred))
  if 'MAX_ERROR' in metrics:
    print("MAX_ERROR: ", max_error(y_test, y_pred))
  if 'RMSE' in metrics:
    print("RMSE: ", np.sqrt(mean_squared_error(y_test, y_pred)))
  if 'R2' in metrics:
    print("R2: ", r2_score(y_test, y_pred))
  if 'MAPE' in metrics:
    # используем метрику MAPE только при отсутсвии нулей
    if make_non_zero:
      mask = y_test != 0
      y_test = y_test[mask]
      y_pred = y_pred[mask]
      print("MAPE: ", mean_absolute_percentage_error(y_test, y_pred))
```

Для удобной работы с метриками и вычисления их разницы между разними методами добавим представление метрик в виде класса.

```python

class Metrics():
  def __init__(self, y_test,
               y_pred,
               metrics=['MSE', 'MAE', 'MAX_ERROR', 'RMSE', 'R2', 'MAPE'],
               ) -> None:
    self.y_test = y_test
    self.y_pred = y_pred

    self.metrics = metrics
    #self.make_non_zero = make_non_zero

    #метрики
    self.MSE = None
    self.MAE = None
    self.MAX_ERROR = None
    self.RMSE = None
    self.R2 = None
    self.MAPE = None


  def get_metrics(self, make_non_zero=False):
    if 'MSE' in self.metrics:
      self.MSE = mean_squared_error(self.y_test, self.y_pred)
    if 'MAE' in self.metrics:
      self.MAE = mean_absolute_error(self.y_test, self.y_pred)
    if 'MAX_ERROR' in self.metrics:
      self.MAX_ERROR = max_error(self.y_test, self.y_pred)
    if 'RMSE' in self.metrics:
      self.RMSE = np.sqrt(mean_squared_error(self.y_test, self.y_pred))
    if 'R2' in self.metrics:
      self.R2 = r2_score(self.y_test, self.y_pred)
    if 'MAPE' in self.metrics:
      # используем метрику MAPE только при отсутсвии нулей
      if make_non_zero:
        mask = self.y_test != 0
        y_test_ = self.y_test[mask]
        y_pred_ = self.y_pred[mask][:len(y_test)]
        self.MAPE = mean_absolute_percentage_error(y_test_, y_pred_)

    return self

  def get_dict(self):
    # подготовит словарь для дальнейшего преобразования в DataFrame и опрятного вывода
    return {"MSE": self.MSE,
            "MAE": self.MAE,
            "MAX_ERROR": self.MAX_ERROR,
            "RMSE": self.RMSE,
            "R2": self.R2,
            "MAPE": self.MAPE}

  def __str__(self):
    return f" MSE: {self.MSE} \n MAE: {self.MAE} \n MAX_ERROR: {self.MAX_ERROR} \n RMSE: {self.RMSE} \n R2: {self.R2} \n MAPE: {self.MAPE}"
```

Проверка на реализованных моделях.

Такие типы имеют наши признаки. Сейчас важно разделить по сути бинарные признаки от численных перед применением Scaler. Это делается потому, что бинарные кодируют по сути Есть/Нет и "скалировать" их незачем.

```python
def get_bin_cols(X_train):
  bin_cols = []
  # признаки значения которых бинарны
  for col in X_train.columns:
    contains = X_train[col].unique()
    if set(contains).issubset({0, 1}):
      bin_cols.append(col)
  return bin_cols
```

Тогда все остальные "пригодны" для Scaler

```python
def get_num_cols(X_train):
  num_cols = []
  for col in X_train.columns:
    if col not in get_bin_cols(X_train):
      num_cols.append(col)
  return num_cols
```

```python
num_cols = get_num_cols(X_train)
num_cols
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# 1. Базовый датасет без трансформаций
X_train_base = X_train.copy()
X_test_base = X_test.copy()

# 2. StandartScaler
scaler = StandardScaler()

X_train_scale = X_train.copy()
X_test_scale = X_test.copy()
# бинарные не скейлим
X_train_scale[num_cols] = scaler.fit_transform(X_train[num_cols])
# без fit для тестовой выборки
X_test_scale[num_cols] = scaler.transform(X_test[num_cols])

```
